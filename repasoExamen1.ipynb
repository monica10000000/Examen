2.  ¿Cuántos tickets se registraron en cada sucursal? ¿Cuál registró más?



```python
sucursalSur = len (dfVentas[dfVentas ['sucursal']=='Sur'])
sucursalNorte = len (dfVentas[dfVentas ['sucursal']=='Norte'])
sucursalCentro = len (dfVentas[dfVentas ['sucursal']=='Centro'])

print('En la sucursal Sur hay', sucursalSur, 'tickets')
print('En la sucursal Norte hay', sucursalNorte, 'tickets')
print('En la sucursal Centro hay', sucursalCentro, 'tickets')

if sucursalSur > sucursalNorte and sucursalSur > sucursalCentro:
    print (f"La sucursal Sur tiene {sucursalSur} tickets")
elif sucursalNorte > sucursalSur and sucursalNorte > sucursalCentro:
    print (f"La sucursal Norte tiene {sucursalNorte} tickets")
elif sucursalCentro > sucursalSur and sucursalCentro > sucursalNorte:
    print (f"La sucursal Centro tiene {sucursalCentro} tickets")
else:
    print("Hay un empate entre las sucursales")
```

3.¿Cuántos tickets se pagaron con cada método y cuántos tienen ese dato sin registrar?

```python
transferencia = len(dfVentas[dfVentas['metodo_pago'] == 'Transferencia'])
efectivo =  len(dfVentas[dfVentas['metodo_pago'] == 'Efectivo'])
tarjeta = len(dfVentas[dfVentas['metodo_pago'] == 'Tarjeta'])
sinRegistrar = len(dfVentas) - (transferencia+ efectivo+ tarjeta)

print (transferencia, 'tickets se pagáron con tranferencia')
print (efectivo, 'tickets se pagáron con efectivo')
print (sinRegistrar, 'tickets se pagáron con sin registrar')
```
